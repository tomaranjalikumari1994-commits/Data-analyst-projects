## Named Entity Recognition (NER)
    NER helps you extract entities like:

    Person Names (PERSON)

    Organizations (ORG)

    Locations (GPE, LOC)

    Dates, Time, Money, etc.

```python
import spacy

# Load small English model
nlp = spacy.load("en_core_web_sm")
```

```python
text = """
Apple Inc. was founded by Steve Jobs, Steve Wozniak, and Ronald Wayne in April 1976.
The company is headquartered in Cupertino, California.
Tim Cook is the current CEO.
Apple's market valuation crossed 3 trillion USD in 2022.
"""
```

    doc.ents gives named entities found in the text.
    ent.text is the word or phrase.
    ent.label_ is the type of entity: PERSON, ORG, GPE, etc.



```python
# Pass the text through the NLP pipeline
doc = nlp(text)

# Extract entities
for ent in doc.ents:
    print(f"{ent.text} --> {ent.label_}")
```
