```python
from transformers import pipeline

#Load summarizer pipeline

summarizer = pipeline("summarization",model="facebook/bart-Large-cnn")
```

```python
article = """
The global tech industry has experienced rapid growth over the past decade.
Major players like Google, Amazon, Apple, and Microsoft have expanded their reach into emerging technologies
such as Artificial Intelligence, Cloud Computing, and the Internet of Things. Startups across the world are also contributing to innovation at scale.
However, challenges like data privacy, regulatory compliance, and talent shortages continue to pose risks for sustained growth.
"""
```

    pipeline("summarization"): Loads a model trained for abstractive summarization.
    max_length and min_length: Control how short/long the summary will be.
    do_sample=False: Makes output deterministic.

```python
summary=summarizer(article,max_length=60,min_length=20,do_sample=False)
print("summary:\n",summary[0]['summary_text'])
```
